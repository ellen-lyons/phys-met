Comparing weather variables from Chandra to the climatological averages. My assigned variable is temperature.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

VARIABLE = 'wetb'

# Ordered south -> north. Check the file names match your downloads.
STATIONS = {
    'Valentia Island':   '/home/ellen/git projects/phys-met/DATA/hly2275.csv',
    'Dublin Airport': '/home/ellen/git projects/phys-met/DATA/hly532.csv',
    'Malin Head':     '/home/ellen/git projects/phys-met/DATA/hly1575.csv',
}

STORM_START = '2026-01-26 00:00'
STORM_END   = '2026-01-28 12:00'
CENTRE_DAY  = pd.Timestamp('2026-01-26')   # window 11 Jan - 10 Feb, as on Sara's slide
WINDOW      = 15                           # +- days
QUANTILES   = [0.05, 0.25, 0.5, 0.75, 0.95]

LABELS = {
    'temp':  ('Air temperature', 'T2m (°C)'),
    'dewpt': ('Dew point', 'Td (°C)'),
    'wetb':  ('Wet-bulb temperature', 'Tw (°C)')
    #'depr':  ('Dew point depression (T − Td)', 'T − Td (°C)'),
}
```

```python
# ---- Functions --------------------------------------------------------------
def read_station(path):
    """Read a Met Eireann hourly file. The text header length differs between
    stations, so find the column-name row instead of hard-coding skiprows."""
    with open(path) as f:
        for i, line in enumerate(f):
            if line.startswith('date,'):
                break
    df = pd.read_csv(path, skiprows=i, usecols=['date', 'temp', 'wetb', 'dewpt'],
                     index_col='date', parse_dates=['date'],
                     date_format='%d-%b-%Y %H:%M',
                     low_memory=False)  # blank values mix types; read in one go
    df = df.apply(pd.to_numeric, errors='coerce')            # blanks -> NaN
    df = df[df.index.notna()].sort_index()                    # drop unparsed dates
    #df['depr'] = df['temp'] - df['dewpt']
    return df


def hourly_climatology(series, centre, window=WINDOW):
    """Percentiles for each hour of day, 1991-2020, +-window days.
    Returns a DataFrame indexed by hour (0-23), one column per quantile."""
    clim = series.loc['1991-01-01':'2020-12-31']
    doy = centre.dayofyear
    diff = np.abs(clim.index.dayofyear - doy)
    year_len = np.where(clim.index.is_leap_year, 366, 365)
    circ_diff = np.minimum(diff, year_len - diff)
    clim_window = clim[circ_diff <= window]
    print(f"   climatology: {clim_window.notna().sum()} hourly values in window")
    return clim_window.groupby(clim_window.index.hour).quantile(QUANTILES).unstack()
```

```python
# plot
long_name, ylabel = LABELS[VARIABLE]
fig, axes = plt.subplots(len(STATIONS), 1, figsize=(9, 8), sharex=True)

for ax, (name, path) in zip(axes, STATIONS.items()):
    print(name)
    data = read_station(path)[VARIABLE]
    clim = hourly_climatology(data, CENTRE_DAY)

    obs = data.loc[STORM_START:STORM_END]
    # Repeat the 24-hour climatology onto every hour of the storm period
    c = clim.reindex(obs.index.hour)
    c.index = obs.index

    ax.fill_between(c.index, c[0.05], c[0.95], color='red', alpha=0.12, label='p05–p95')
    ax.fill_between(c.index, c[0.25], c[0.75], color='red', alpha=0.3, label='p25–p75')
    ax.plot(c.index, c[0.5], color='red', lw=1.5, label='p50')
    ax.plot(obs.index, obs, color='black', marker='o', ms=3, ls=':', label='obs')

    n = obs.notna().sum()
    above75, above95 = (obs > c[0.75]).sum(), (obs > c[0.95]).sum()
    below25, below05 = (obs < c[0.25]).sum(), (obs < c[0.05]).sum()
    ax.set_title(f'{name}: {above75}/{n} h above p75 ({above95} above p95), '
                 f'{below25}/{n} h below p25 ({below05} below p05)',
                 loc='left', fontsize=9)
    ax.set_ylabel(ylabel)
    ax.grid(alpha=0.4)

    print(f"   storm period: max {obs.max():.1f}, min {obs.min():.1f} | "
          f">p75 {above75}, >p95 {above95}, <p25 {below25}, <p05 {below05} (of {n} h)")

axes[0].legend(loc='center left', bbox_to_anchor=(1.01, 0.5), fontsize=9)
axes[-1].xaxis.set_major_locator(mdates.HourLocator(byhour=[0, 6, 12, 18]))
axes[-1].xaxis.set_major_formatter(mdates.DateFormatter('%d %b\n%H:%M'))
axes[-1].set_xlabel('Time (UTC)')
fig.suptitle(f'Storm Chandra: {long_name} vs 1991–2020 climatology (±15 days)')

plt.tight_layout()
plt.savefig(f'chandra_{VARIABLE}_climate.png', dpi=200, bbox_inches='tight')
plt.show()
```

In the code above, I plotted temp, wet-bulb, dew-point, and had a look at how they vary compared to the averages. Temperature is pretty boring and average, as expected, but dew point/wet-bulb are more interesting, especially at Dublin Airport.

Below I will focus in on dew-point temp at Dublin and find the anomaly: observed dew point - climate median for that hour of day!

```python
STATION = 'Dublin Airport'
FILE = '/home/ellen/git projects/phys-met/DATA/hly532.csv'
MARK_LOW = False

# ---- Read data --------------------------------------------------------------
# The text header length differs between stations, so find the column-name row.
with open(FILE) as f:
    for i, line in enumerate(f):
        if line.startswith('date,'):
            break
df = pd.read_csv(FILE, skiprows=i, usecols=['date', 'dewpt', 'msl'],
                 index_col='date', parse_dates=['date'],
                 date_format='%d-%b-%Y %H:%M', low_memory=False)
df = df.apply(pd.to_numeric, errors='coerce')   # blanks -> NaN
df = df[df.index.notna()].sort_index()
```

```python
# ---- Climatology ------------------------------------------------------------
clim = df.loc['1991-01-01':'2020-12-31', 'dewpt']
doy = CENTRE_DAY.dayofyear
diff = np.abs(clim.index.dayofyear - doy)
year_len = np.where(clim.index.is_leap_year, 366, 365)
circ_diff = np.minimum(diff, year_len - diff)
clim_window = clim[circ_diff <= WINDOW]
pct = clim_window.groupby(clim_window.index.hour).quantile([0.25, 0.5, 0.75]).unstack()

storm = df.loc[STORM_START:STORM_END]
obs = storm['dewpt']
c = pct.reindex(obs.index.hour)      # repeat 24-h climatology for each day
c.index = obs.index

anom = obs - c[0.5]
p25_anom = c[0.25] - c[0.5]          # normal range, as anomalies from the median
p75_anom = c[0.75] - c[0.5]

t_minp = storm['msl'].idxmin()
before, after = anom.loc[:t_minp], anom.loc[t_minp:]

print(f"Climatology values in window: {clim_window.notna().sum()}")
print(f"Minimum MSL {storm['msl'].min():.1f} hPa at {t_minp:%d %b %H:%M}")
print(f"Before the low: mean anomaly {before.mean():+.1f} °C, "
      f"max {before.max():+.1f} °C at {before.idxmax():%d %b %H:%M}")
print(f"After the low:  mean anomaly {after.mean():+.1f} °C, "
      f"min {after.min():+.1f} °C at {after.idxmin():%d %b %H:%M}")
print(f"Hours above p75: {(obs > c[0.75]).sum()}, below p25: {(obs < c[0.25]).sum()} "
      f"(of {obs.notna().sum()})")
```

```python
plt.rcParams.update({
    'font.size': 14, 'axes.titlesize': 16, 'axes.labelsize': 16,
    'xtick.labelsize': 12, 'ytick.labelsize': 14, 'legend.fontsize': 13,
    'font.family': 'sans-serif',
    'font.sans-serif': ['Arial', 'Helvetica', 'DejaVu Sans'],
    'mathtext.fontset': 'dejavusans',
})

# ---- Plot -------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(11,5))

ax.fill_between(anom.index, 0, anom, where=anom >= 0, interpolate=True,
                color='firebrick', alpha=0.7, label='moister than normal')
ax.fill_between(anom.index, 0, anom, where=anom < 0, interpolate=True,
                color='steelblue', alpha=0.7, label='drier than normal')
ax.plot(anom.index, anom, color='black', lw=1)
ax.plot(p75_anom.index, p75_anom, color='grey', ls='--', lw=1, label='normal range (p25–p75)')
ax.plot(p25_anom.index, p25_anom, color='grey', ls='--', lw=1)
ax.axhline(0, color='black', lw=0.8)

if MARK_LOW:
    # Time the low passed Dublin (lowest MSL), as a reference to the pressure slide
    ax.axvline(t_minp, color='black', ls=':', lw=1.2)
    ax.text(t_minp, ax.get_ylim()[1], f' low passes\n {t_minp:%d %b %H:%M}',
            va='top', fontsize=8)

ax.set_ylabel('Dew point anomaly (°C)')
#ax.set_xlabel('Date & Time')
ax.set_title('Dew Point Anomaly at Dublin Airport vs Climatological Median')
ax.xaxis.set_major_locator(mdates.HourLocator(byhour=[0, 6, 12, 18]))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%d %b %H:%M'))
plt.setp(ax.get_xticklabels(), rotation=45, ha='right')
ax.legend(loc='center left', bbox_to_anchor=(1.01, 0.5), frameon=False)
ax.grid(alpha=0.3)

plt.tight_layout()
#plt.savefig('chandra_dewpoint_anomaly.png', dpi=200, bbox_inches='tight', transparent=True)
plt.show()
```
